## 1.2. LLM sử dụng

````python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
from transformers.models.gemma3 import Gemma3ForConditionalGeneration
from abc import ABC, abstractmethod

class BaseLLM(ABC):
    @abstractmethod
    def generate_batch(self, system_prompt: str, user_prompts: list[str], max_new_tokens: int = 256, **gen_kwargs) -> list[str]:
        pass

class QwenItService(BaseLLM):
    def __init__(self, model_name='/kaggle/input/models/qwen-lm/qwen2.5/transformers/7b-instruct/1', device='cuda'):
        self.device = device
        self.tokenizer = AutoTokenizer.from_pretrained(model_name)

        # set padding token
        self.tokenizer.padding_side = 'left'
        if self.tokenizer.pad_token is None:
            self.tokenizer.pad_token = self.tokenizer.eos_token

        print("Đang nạp trọng số model lên VRAM...")
        self.model = AutoModelForCausalLM.from_pretrained(
            model_name,
            dtype=torch.float16,
            device_map="auto"
        )
        print("Đã tải xong model!")

    def generate_batch(self, system_prompt: str, user_prompts: list[str], max_new_tokens: int = 256, **gen_kwargs) -> list[str]:
        # Chuẩn bị list đoạn chat đầu vào
        texts_chat = []
        for prompt in user_prompts:
            messages = [
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": prompt}
            ]
            text = self.tokenizer.apply_chat_template(
                messages,
                tokenize=False,
                add_generation_prompt=True
            )
            texts_chat.append(text)
        # Tokenize với Padding
        model_inputs = self.tokenizer(
            texts_chat,
            return_tensors="pt",
            padding=True,
            truncation=True
        ).to(self.device)

        if any(k in gen_kwargs for k in ['temperature', 'top_k', 'top_p']):
            gen_kwargs['do_sample'] = gen_kwargs.get('do_sample', True)
        # 3. Generate
        generated_ids = self.model.generate(
            **model_inputs,
            max_new_tokens=max_new_tokens,
            pad_token_id=self.tokenizer.pad_token_id,
            **gen_kwargs
        )
        # 4. Cắt bỏ phần prompt (Slicing)
        generated_ids = [
            output_ids[len(input_ids):] for input_ids, output_ids in zip(model_inputs.input_ids, generated_ids)
        ]

        responses = self.tokenizer.batch_decode(generated_ids, skip_special_tokens=True)
        return responses
````

## 1.4. Reranker


````python
import torch
from transformers import AutoModelForSequenceClassification, AutoTokenizer

class BaseRerankerModel(ABC):
    @abstractmethod
    def score(self, query, documents):
        pass

class BGEReranker(BaseRerankerModel):
    def __init__(self, model_name='/kaggle/input/models/andreasbis/baai-bge-reranker-v2-m3/transformers/default/1', device='cpu'):
        self.device = device
        self.tokenizer = AutoTokenizer.from_pretrained(model_name)
        self.model = AutoModelForSequenceClassification.from_pretrained(model_name, device_map="auto")
        self.model.eval()

    def score(self, query, documents):
        pairs = [[query, doc] for doc in documents]
        with torch.no_grad():
            inputs = self.tokenizer(pairs, padding=True, truncation=True, return_tensors='pt', max_length=512).to(self.device)
            scores = self.model(**inputs, return_dict=True).logits.view(-1, ).float()
        return scores.cpu().tolist()
````

## 1.5. Xử lý dữ liệu đầu vào

````python
import re
from collections import deque

class DataProcessor:
    def __init__(self, tokenizer, sentence_splitter):
        self.tokenizer = tokenizer
        self.sentence_splitter = sentence_splitter

    @staticmethod
    def clean_text(text):
        """
        Load and preprocess the context data for chunking
        """
        # Chuẩn hóa xuống dòng và xóa khoảng trắng
        text = re.sub(r'\n\s*\n', '\n\n', text)
        text = re.sub(r'[ \t]+', ' ', text)

        # Xóa kí tự đặc biệt
        text = re.sub(r'[^\w\s\.,;:!?\-\(\)\[\]{}"\']', '', text)
        return text.strip()

    def process_text_to_chunks(self, text, chunk_token_size, chunk_start_idx):
        cleaned_text = DataProcessor.clean_text(text)
        if not cleaned_text:
            return []

        sentences = sentence_splitter.tokenize(cleaned_text)
        if not sentences:
            return []

        # Tokenize tất cả câu 1 lần duy nhất
        batch_encoded = self.tokenizer(
            sentences,
            add_special_tokens=False,
            truncation=False,
            return_attention_mask=False,
        )
        token_list = batch_encoded["input_ids"]  # đây là list[list]

        queue = deque(zip(sentences, token_list))

        all_chunks = []
        chunk_idx = chunk_start_idx

        while queue:
            chunk_pairs = []
            total_tokens = 0

            while queue:
                sent, tokens = queue.popleft()
                chunk_pairs.append((sent, tokens))
                total_tokens += len(tokens)

                if total_tokens >= chunk_token_size:
                    break

            all_chunks.append({
                "text": " ".join(s for s, _ in chunk_pairs),
                "chunk_index": chunk_idx,
                "token_count": total_tokens,
            })
            chunk_idx += 1

        return all_chunks
````

## 1.6. Trích xuất thực thể, quan hệ

````python
class Extractor:
    def __init__(self, llm: BaseLLM):
        self.llm = llm

    def _parse_response_data(self, text, chunk_index):
        entities = []
        relations = []
        content_keywords = []

        TUPLE_DEL = PROMPTS["DEFAULT_TUPLE_DELIMITER"]   # "<|>"
        RECORD_DEL = PROMPTS["DEFAULT_RECORD_DELIMITER"]  # "##"
        COMPLETION = PROMPTS["DEFAULT_COMPLETION_DELIMITER"]

        text = text.replace(COMPLETION, "").strip()
        
        # thay \n bằng ## nếu ## không xuất hiện
        if RECORD_DEL not in text:
            text = text.replace("\n", RECORD_DEL)
        else:
            # Có thể vừa có ## vừa có \n, chuẩn hóa hết
            text = text.replace("\n", RECORD_DEL)

        records = text.split(RECORD_DEL)

        for record in records:
            record = record.strip()
            if not record:
                continue

            if record.startswith("("):
                record = record[1:]
            if record.endswith(")"):
                record = record[:-1]

            # Bỏ qua record bị truncate
            parts = [p.strip().strip('"') for p in record.split(TUPLE_DEL)]

            if len(parts) < 2:
                continue

            tag = parts[0].lower()

            if tag == "entity" and len(parts) >= 4:
                entities.append({
                    "name": parts[1],
                    "type": parts[2],
                    "description": parts[3],
                })
            elif tag == "relationship" and len(parts) >= 5:
                relations.append({
                    "source": parts[1],
                    "target": parts[2],
                    "description": parts[3],
                    "keywords": parts[4] if len(parts) > 4 else "",
                    "strength": parts[5] if len(parts) > 5 else "",
                })
            elif tag == "content_keywords" and len(parts) >= 2:
                content_keywords.append({
                    "keywords": [kw.strip() for kw in parts[1].split(",")]
                })

        for entity_id, entity in enumerate(entities):
            entity["id"] = f"{chunk_index}_e{entity_id}"

        name_to_id = {e["name"]: e["id"] for e in entities}
        for relation_id, relation in enumerate(relations):
            relation["id"] = f"{chunk_index}_r{relation_id}"
            relation["source_id"] = name_to_id.get(relation["source"], None)
            relation["target_id"] = name_to_id.get(relation["target"], None)

        return {"entities": entities, "relations": relations, "content_keywords": content_keywords}

    def parse_batch(self, responses, chunk_index: list[int]):
        batch_results = []
        for response, chunk_idx in zip(responses, chunk_index):
            batch_results.append(self._parse_response_data(response, chunk_idx))

        return batch_results


    def extract_batch(self, input_texts: list[str], chunk_index: list[int]):
        examples_str = "\n".join([
            ex.format(
                tuple_delimiter=PROMPTS["DEFAULT_TUPLE_DELIMITER"],
                record_delimiter=PROMPTS["DEFAULT_RECORD_DELIMITER"],
                completion_delimiter=PROMPTS["DEFAULT_COMPLETION_DELIMITER"]
            ) for ex in PROMPTS["entity_extraction_examples"]
        ])

        system_prompt = PROMPTS["entity_relation_extraction_system_prompt"].format(
            entity_types=", ".join(PROMPTS["DEFAULT_ENTITY_TYPES"]),
            tuple_delimiter=PROMPTS["DEFAULT_TUPLE_DELIMITER"],
            record_delimiter=PROMPTS["DEFAULT_RECORD_DELIMITER"],
            completion_delimiter=PROMPTS["DEFAULT_COMPLETION_DELIMITER"],
            examples=examples_str,
            language="English",
        )

        user_prompts = []
        for input_text in input_texts:
            user_prompt = PROMPTS["entity_relation_extraction_user_prompt"].format(
                language="English",
                input_text=input_text,
            )
            user_prompts.append(user_prompt)

        responses = self.llm.generate_batch(
            system_prompt,
            user_prompts,
            max_new_tokens=2048,
            do_sample=False,
        )

        return responses
````

## 1.7. Kiểm tra lại kết quả trả về từ Extractor

````python
import re
INVALID_CHARS = re.compile(r"[<>{}\[\]|\\]")  # delimiter / artifact chars

class Validator:
    def __init__(self):
        pass

    @staticmethod
    def is_valid_entity(e: dict) -> bool:
        required_fields = ["id", "name", "type", "description"]
    
        # 1. Kiểm tra các trường bắt buộc có bị thiếu hay không
        if not all(f in e for f in required_fields):
            return False
    
        entity_id = e["id"]
        name = e["name"]
        etype = e["type"]
        desc = e["description"]
    
        # 2. Tất cả các trường phải có kiểu chuỗi
        if not all(isinstance(x, str) for x in [entity_id, name, etype, desc]):
            return False
    
        # 3. Không được để trống hoặc chỉ chứa khoảng trắng
        if not name.strip() or not etype.strip() or not desc.strip():
            return False
    
        # 4. Loại bỏ các ký tự đặc biệt hoặc lỗi định dạng
        for field in [entity_id, name, etype, desc]:
            if INVALID_CHARS.search(field):
                return False
    
        # 5. Kiểm tra định dạng của ID
        if not re.match(r"^\d+_e\d+$", entity_id):
            return False
    
        # 6. Kiểm tra độ dài hợp lệ của tên thực thể
        if len(name) < 2 or len(name) > 200:
            return False
    
        # 7. Kiểm tra độ dài tối thiểu của mô tả
        if len(desc) < 10:
            return False
    
        return True

    @staticmethod
    def is_valid_relation(r: dict, valid_entity_ids: set) -> bool:
        required_fields = [
            "id", "source", "target",
            "source_id", "target_id",
            "description", "keywords", "strength"
        ]
    
        # 1. Kiểm tra các trường bắt buộc có bị thiếu hay không
        if not all(f in r for f in required_fields):
            return False
    
        # 2. Kiểm tra tất cả các trường đều có kiểu chuỗi
        if not all(isinstance(r[f], str) for f in required_fields):
            return False
    
        # 3. Không được để trống hoặc chỉ chứa khoảng trắng
        if any(not r[f].strip() for f in required_fields):
            return False
    
        # 4. Kiểm tra định dạng của ID
        if not re.match(r"^\d+_r\d+$", r["id"]):
            return False
    
        # 5. Kiểm tra thực thể nguồn và đích có tồn tại trong tập thực thể hay không
        if r["source_id"] not in valid_entity_ids:
            return False
        if r["target_id"] not in valid_entity_ids:
            return False
    
        # 6. Kiểm tra độ dài tối thiểu của mô tả
        if len(r["description"]) < 10:
            return False
    
        # 7. Kiểm tra độ dài hợp lệ của từ khóa
        if len(r["keywords"]) > 100:
            return False
    
        # 8. Kiểm tra trọng số phải là số nguyên trong khoảng từ 1 đến 10
        try:
            strength = int(r["strength"])
        except:
            return False
    
        if not (1 <= strength <= 10):
            return False
    
        # 9. Không cho phép quan hệ tự nối (self-loop)
        if r["source_id"] == r["target_id"]:
            return False
    
        # 10. Loại bỏ các ký tự đặc biệt hoặc lỗi định dạng
        for field in ["source", "target", "description", "keywords"]:
            if INVALID_CHARS.search(r[field]):
                return False
    
        return True
````

## 1.8. Loại bỏ trùng lặp

````python

import numpy as np
from sentence_transformers import util
from collections import defaultdict
from difflib import SequenceMatcher
import re

class Deduplicator:
    def __init__(self, embedding_model):
        self.typed_entities_group = None
        self.all_embeddings_pairs = None
        self.canonical_entities = []
        self.final_relations = []
        self.id_to_canonical = None
        self.embedding_model = embedding_model
        
    @staticmethod
    def _normalize_text(text: str) -> str:
        if text is None:
            return ""
    
        text = Deduplicator._strip_parenthetical_acronym(str(text))
        text = text.strip().lower()
        text = re.sub(r"[^a-z0-9]+", " ", text)
        text = re.sub(r"\s+", " ", text).strip()
    
        return text

    @staticmethod
    def _strip_parenthetical_acronym(text: str) -> str:
        """
        Bỏ acronym trong ngoặc.
        Ví dụ:
            basal cell carcinoma (BCC) -> basal cell carcinoma
        """
        if text is None:
            return ""
    
        text = str(text)
    
        # Chỉ bỏ ngoặc nếu nội dung trong ngoặc là acronym ngắn
        text = re.sub(r"\(([A-Z]{2,10})\)", " ", text)
    
        return text

    @staticmethod
    def _extract_parenthetical_acronyms(text: str) -> set:
        """
        Lấy acronym trong ngoặc.
        Ví dụ:
            basal cell carcinoma (BCC) -> {"bcc"}
        """
        if text is None:
            return set()
    
        matches = re.findall(r"\(([A-Za-z]{2,10})\)", str(text))
    
        return {
            Deduplicator._normalize_text(m)
            for m in matches
            if m
        }
        
    @staticmethod
    def _acronym_of(text: str) -> str:
        norm = Deduplicator._normalize_text(text)
        tokens = norm.split()
    
        if not tokens:
            return ""
    
        return "".join(token[0] for token in tokens if token)
    
    
    @staticmethod
    def _build_cluster_object(fuzzy_members):
        """
        Đưa một nhóm entity (members_data) sang cluster object
        để sử dụng trong build_canonical_entity().
        
        Mỗi phần tử trong members_data có dạng:
        
        {
            "entity": ent,
            "chunk_index": chunk_index
        }
        
        Cluster object trả về chứa:
        - medoid đại diện cho cluster
        - danh sách members
        - thông tin chunk_index tương ứng
        """
        members = []
    
        for item in fuzzy_members:
            ent = item["entity"]
            members.append({
                "ent": ent,
                "chunk_index": item["chunk_index"],
                "name_embedding": None,
                "des_embedding": None,
                "dedup_method": "fuzzy",
            })
    
        # Chọn medoid đơn giản:
        # tên dài nhất thường là tên đầy đủ nhất
        medoid_member = max(
            members,
            key=lambda m: len(Deduplicator._normalize_text(m["ent"].get("name", "")).split())
        )
    
        return {
            "medoid": medoid_member["ent"],
            "medoid_chunk_index": medoid_member["chunk_index"],
            "members": members,
            "dedup_method": "fuzzy",
        }

    @staticmethod
    def _exact_acronym_cluster_group(group):
        """
        Cluster bằng:
        - normalize_text()
        - acronym dictionary
    
        O(n)
        """
    
        buckets = defaultdict(list)
        acronym_map = {}
    
        for ent_data in group:
            name = ent_data["entity"].get("name", "")
    
            norm = Deduplicator._normalize_text(name)
    
            buckets[norm].append(ent_data)

            acr = Deduplicator._acronym_of(norm)
    
            if len(acr) >= 2:
                acronym_map[acr] = norm
    
            for p_acr in Deduplicator._extract_parenthetical_acronyms(name):
                acronym_map[p_acr] = norm
    
    
        for key in list(buckets.keys()):
    
            if key in acronym_map:
    
                full_name_key = acronym_map[key]
    
                if full_name_key != key:
    
                    buckets[full_name_key].extend(
                        buckets[key]
                    )
    
                    del buckets[key]
    
        clusters = []
        singletons = []
    
        for members in buckets.values():
    
            if len(members) >= 2:
                clusters.append(members)
            else:
                singletons.append(members[0])
    
        return clusters, singletons
        

    def group_by_type(self, chunks):
        self.typed_entities_group = defaultdict(list)
        for chunk in chunks:
            for ent in chunk["entities"]:
                ent_data = {
                    "entity": ent,
                    "chunk_index": chunk["chunk_index"]
                }
                self.typed_entities_group[ent["type"]].append(ent_data)

    def build_embeddings_pairs(self):
        """
        Embedding sẽ được tạo sau (lazy build) trong deduplicate_entity()
        sau khi fuzzy clustering loại bớt entity đã gộp.
        """
        self.all_embeddings_pairs = None

    @staticmethod
    def _custom_similarity_matrix(group, name_embeddings, des_embeddings,
                              w_name=0.7, w_des=0.3):
        """
        Tính ma trận similarity:
        sim(i,j) = sim(name_i, name_j) * w_name + sim(des_i, des_j) * w_des
        """
        n = len(group)

        # Cosine sim matrix cho name và description
        sim_name = util.cos_sim(name_embeddings, name_embeddings).cpu().numpy()  # (n, n)
        sim_des  = util.cos_sim(des_embeddings,  des_embeddings ).cpu().numpy()  # (n, n)

        combined = w_name * sim_name + w_des * sim_des  # (n, n)
        return combined

    @staticmethod
    def _community_detection_custom(sim_matrix, threshold=0.9):
        """
        Community detection bằng greedy clustering trên custom sim matrix.
        Tương tự logic của sentence_transformers.util.community_detection.
        """
        n = sim_matrix.shape[0]
        # Sắp xếp các cặp theo similarity giảm dần
        pairs = []
        for i in range(n):
            for j in range(i+1, n):
                if sim_matrix[i][j] >= threshold:
                    pairs.append((i, j, sim_matrix[i][j]))
        pairs.sort(key=lambda x: x[2], reverse=True)

        # Union-Find để gộp cluster
        parent = list(range(n))
        def find(x):
            while parent[x] != x:
                parent[x] = parent[parent[x]]
                x = parent[x]
            return x
        def union(x, y):
            parent[find(x)] = find(y)

        for i, j, sim in pairs:
            union(i, j)

        # Gom thành clusters
        clusters_dict = defaultdict(list)
        for idx in range(n):
            clusters_dict[find(idx)].append(idx)

        clusters = sorted(clusters_dict.values(), key=len, reverse=True)
        return clusters

    @staticmethod
    def _compute_centroid(embeddings):
        return np.mean(np.stack(embeddings), axis=0)

    @staticmethod
    def _custom_sim(emb_name_a, emb_des_a, emb_name_b, emb_des_b, w_name=0.7, w_des=0.3):
        """Tính weighted similarity giữa 2 entity."""
        sim_name = util.cos_sim(emb_name_a.reshape(1,-1), emb_name_b.reshape(1,-1)).item()
        sim_des  = util.cos_sim(emb_des_a.reshape(1,-1),  emb_des_b.reshape(1,-1)).item()
        return w_name * sim_name + w_des * sim_des

    @staticmethod
    def _select_medoid(members, w_name=0.7, w_des=0.3):
        """Chọn member gần centroid nhất theo custom similarity."""
        if len(members) == 1:
            return members[0]

        # Tính centroid riêng cho name và description
        centroid_name = Deduplicator._compute_centroid([m['name_embedding'] for m in members])
        centroid_des  = Deduplicator._compute_centroid([m['des_embedding']  for m in members])

        sims = [
            Deduplicator._custom_sim(centroid_name, centroid_des,
                       m['name_embedding'], m['des_embedding'],
                       w_name, w_des)
            for m in members
        ]
        return members[int(np.argmax(sims))]

    @staticmethod
    def _filter_cluster_by_medoid(members, threshold=0.85, w_name=0.7, w_des=0.3):
        """Lọc outlier dựa trên custom similarity với medoid."""
        if len(members) == 1:
            return members, []

        medoid = Deduplicator._select_medoid(members, w_name, w_des)
        core, outliers = [], []

        for member in members:
            sim = Deduplicator._custom_sim(
                medoid['name_embedding'], medoid['des_embedding'],
                member['name_embedding'], member['des_embedding'],
                w_name, w_des
            )
            if sim >= threshold:
                core.append(member)
            else:
                outliers.append({**member, 'outlier_sim': round(sim, 4)})

        return core, outliers

    @staticmethod
    def _select_canonical_name(members):
        """Tên đầy đủ nhất = nhiều token nhất sau normalize."""
        return max(
            members,
            key=lambda m: len(
                Deduplicator._normalize_text(m["ent"].get("name", "")).split()
            )
        )["ent"]["name"]

    @staticmethod
    def _select_top_k_descriptions(members, k=5):
        """top-k description dài nhất, nối bằng ' | '."""
        sorted_members = sorted(members, key=lambda m: len(m['ent']['description']), reverse=True)
        seen, unique_descs = set(), []
        for m in sorted_members[:k]:
            desc = m['ent']['description']
            if desc not in seen:
                seen.add(desc)
                unique_descs.append(desc)
        return ' | '.join(unique_descs)

    @staticmethod
    def _build_canonical_entity(cluster, k=5, w_name=0.7, w_des=0.3):
        members = cluster['members']
        medoid  = cluster['medoid']  
        medoid_chunk_index = cluster['medoid_chunk_index']

        canonical_name = Deduplicator._select_canonical_name(members)
        merged_desc    = Deduplicator._select_top_k_descriptions(members, k=k)

        aliases = [
            { 'ent': m['ent'], 'chunk_index': m['chunk_index'] }
            for m in members
            if m['ent']['id'] != medoid['id']
        ]

        # Gom tất cả chunk_index từ tất cả các members (không trùng lặp)
        all_chunk_indices = list({m["chunk_index"] for m in members})

        self_ent = {
            'id':          medoid['id'],
            'name':        canonical_name,
            'type':        medoid['type'],
            'description': merged_desc,
            'chunk_index': all_chunk_indices, # list of chunk index
        }

        return {
            'self': self_ent,
            'aliases':     aliases,
            '_member_count': len(members),
        }

    def deduplicate_entity(self):
        """
        Deduplicate entity theo 2 tầng.
        
        Tầng 1: Exact match + Acronym matching.
            - Chuẩn hóa tên entity bằng normalize_text().
            - Gom các entity có tên giống nhau sau khi chuẩn hóa.
            - Gom acronym (ví dụ: BCC) với tên đầy đủ tương ứng
              (ví dụ: Basal Cell Carcinoma).
        
        Tầng 2: Embedding clustering cho các entity còn lại.
            - Chỉ encode các entity chưa được gộp ở tầng 1.
            - Sử dụng similarity giữa name embedding và
              description embedding để phát hiện các entity tương đồng ngữ nghĩa.
        """   
        EMB_THRESHOLD = 0.75
        W_NAME = 0.8
        W_DES = 0.2
        TOP_K_DESC = 5
    
        canonical_entities = defaultdict(list)
    
        for g_type, group in self.typed_entities_group.items():
            if not group:
                continue
    
            # Exact match + Acronym dictionary
            exact_clusters, remaining_entities = (
                Deduplicator._exact_acronym_cluster_group(group)
            )

            for exact_members in exact_clusters:
                cluster_obj = Deduplicator._build_cluster_object(
                    exact_members
                )
            
                canonical = Deduplicator._build_canonical_entity(
                    cluster_obj,
                    k=TOP_K_DESC,
                    w_name=W_NAME,
                    w_des=W_DES,
                )
            
                canonical["dedup_method"] = "exact_acronym"
            
                canonical_entities[g_type].append(canonical)
                
            # 2. Những entity còn lại mới đi qua embedding clustering
            if not remaining_entities:
                continue
    
            if len(remaining_entities) == 1:
                single = remaining_entities[0]
                cluster_obj = {
                    "medoid": single["entity"],
                    "medoid_chunk_index": single["chunk_index"],
                    "members": [{
                        "ent": single["entity"],
                        "chunk_index": single["chunk_index"],
                        "name_embedding": None,
                        "des_embedding": None,
                        "dedup_method": "singleton",
                    }],
                    "dedup_method": "singleton",
                }
    
                canonical = Deduplicator._build_canonical_entity(
                    cluster_obj,
                    k=TOP_K_DESC,
                    w_name=W_NAME,
                    w_des=W_DES,
                )
                canonical["dedup_method"] = "singleton"
                canonical_entities[g_type].append(canonical)
                continue
    
            # Encode chỉ remaining entities
            names = [
                ent_data["entity"].get("name", "")
                for ent_data in remaining_entities
            ]
            descs = [
                ent_data["entity"].get("description", "")
                for ent_data in remaining_entities
            ]
    
            name_embs = self.embedding_model.encode(names)
            des_embs = self.embedding_model.encode(descs)
    
            sim_matrix = Deduplicator._custom_similarity_matrix(
                remaining_entities,
                name_embs,
                des_embs,
                W_NAME,
                W_DES,
            )
    
            g_clusters = Deduplicator._community_detection_custom(
                sim_matrix,
                threshold=EMB_THRESHOLD,
            )
    
            raw_clusters = []
    
            for cluster_indices in g_clusters:
                members = []
    
                for idx in cluster_indices:
                    members.append({
                        "ent": remaining_entities[idx]["entity"],
                        "chunk_index": remaining_entities[idx]["chunk_index"],
                        "name_embedding": name_embs[idx].cpu().numpy(),
                        "des_embedding": des_embs[idx].cpu().numpy(),
                        "dedup_method": "embedding",
                    })
    
                raw_clusters.append(members)
    
            filtered_clusters = []
    
            for members in raw_clusters:
                core, outliers = Deduplicator._filter_cluster_by_medoid(
                    members,
                    EMB_THRESHOLD,
                    W_NAME,
                    W_DES,
                )
    
                if core:
                    medoid = Deduplicator._select_medoid(core, W_NAME, W_DES)
                    filtered_clusters.append({
                        "medoid": medoid["ent"],
                        "medoid_chunk_index": medoid["chunk_index"],
                        "members": core,
                        "dedup_method": "embedding",
                    })
    
                for outlier in outliers:
                    filtered_clusters.append({
                        "medoid": outlier["ent"],
                        "medoid_chunk_index": outlier["chunk_index"],
                        "members": [outlier],
                        "dedup_method": "embedding_outlier_singleton",
                    })
    
            for cluster in filtered_clusters:
                canonical = Deduplicator._build_canonical_entity(
                    cluster,
                    k=TOP_K_DESC,
                    w_name=W_NAME,
                    w_des=W_DES,
                )
                canonical["dedup_method"] = cluster.get("dedup_method", "embedding")
                canonical_entities[g_type].append(canonical)
    
        self.canonical_entities = []
        for entities in canonical_entities.values():
            self.canonical_entities.extend(entities)

    def build_canonical_lookup(self):
         # Build lookup: old_entity_id → canonical_entity
        self.id_to_canonical = {}  # '0_e0' → canonical entity dict

        for canonical in self.canonical_entities:
            # Map chính bản thân medoid
            self.id_to_canonical[canonical['self']['id']] = canonical
            # Map tất cả aliases
            for alias in canonical['aliases']:
                self.id_to_canonical[alias['ent']['id']] = canonical

    def deduplicate_relation(self, chunks):
        # Merge relations

        merged_relations = {}  # key: (canonical_src_id, canonical_tgt_id) → relation

        for chunk in chunks:
            for rel in chunk.get('relations', []):
                src_canonical = self.id_to_canonical.get(rel['source_id'])
                tgt_canonical = self.id_to_canonical.get(rel['target_id'])

                # Bỏ qua nếu không map được
                if not src_canonical or not tgt_canonical:
                    print(f"[WARN] Cannot map relation {rel['id']}: "
                          f"{rel['source_id']} - {rel['target_id']}")
                    continue

                # Bỏ self-loop (2 entity khác nhau bị gộp thành 1)
                if src_canonical['self']['id'] == tgt_canonical['self']['id']:
                    continue

                edge_key = tuple(sorted([src_canonical['self']['id'], tgt_canonical['self']['id']])) # Vô hướng

                if edge_key not in merged_relations:
                    merged_relations[edge_key] = {
                        'source_id':   src_canonical['self']['id'],
                        'source':      src_canonical['self']['name'],
                        'target_id':   tgt_canonical['self']['id'],
                        'target':      tgt_canonical['self']['name'],
                        'description': rel['description'],
                        'keywords':    rel['keywords'],
                        'strength':      float(rel['strength']),
                        'chunk_index': {chunk['chunk_index']}
                    }
                else:
                    existing = merged_relations[edge_key]
                    # Cộng dồn weight
                    existing['strength'] += float(rel['strength'])
                    # Nối description nếu khác nhau
                    if rel['description'] not in existing['description']:
                        existing['description'] += f" ||| {rel['description']}"
                    # Merge keywords
                    old_kws = set(existing['keywords'].split(', '))
                    new_kws = set(rel['keywords'].split(', '))
                    existing['keywords'] = ', '.join(old_kws | new_kws)
                    # Thêm chunk_idx mới nếu chunk_idx của rel này chưa có
                    existing['chunk_index'].add(chunk['chunk_index'])

        # Convert sang list
        # self.final_relations = list(merged_relations.values())
        self.final_relations = [
            {**rel, "chunk_index": list(rel["chunk_index"])}
            for rel in merged_relations.values()
        ]

````

## 1.9. Knowledge Graph (Graph Store)

````python
import networkx as nx
class GraphStore: 
    def __init__(self):
        self.graph = nx.Graph()
    def add_nodes(self, canonical_entities):
        for ent in canonical_entities:
            self.graph.add_node(
                ent['self']['id'],
                metas = ent
            )
    def add_edges(self, final_relations):
        for rel in final_relations:
            self.graph.add_edge(
                rel['source_id'],
                rel['target_id'],
                metas = rel
            )

    def get_node(self, node_id):
        if node_id not in self.graph.nodes:
            return None
        node = self.graph.nodes[node_id]
        new_node = {**node['metas']['self'], "degree": self.graph.degree(node_id)}
        return new_node


    def get_neighbors(self, node_id):
        nb_datas = []
        for (nb_id, edge) in self.graph[node_id].items():
            src_degree = self.graph.degree(node_id)
            tgt, tgt_degree = self.graph.nodes[nb_id], self.graph.degree(nb_id)
            neighbor = {**tgt['metas']['self'], "degree": tgt_degree}
            relation = {**edge['metas'], "degree": src_degree + tgt_degree}
            nb_data = {"neighbor": neighbor, "relation": relation}
            nb_datas.append(nb_data)
        return nb_datas
````

````python
import faiss
import numpy as np
import json


class VectorStore:
    def __init__(self, embedding_model):
        self.entity_store = None
        self.relation_store = None
        self.embedding_model = embedding_model

    def encode(self, texts):
        embs = self.embedding_model.encode_np(texts)
        embs = np.array(embs)
        embs = embs / (np.linalg.norm(embs, axis=1, keepdims=True) + 1e-12) # normalize cho cosine sim
        return embs.astype(np.float32)


    def build_entity_store(self, canonical_entities):
        entity_docs = []
        entity_ids = []
        entity_metas = []
        for ent in canonical_entities:
            doc = f"{ent['self']['name']}"
            entity_docs.append(doc)
            entity_ids.append(ent['self']['id'])
            entity_metas.append(ent)

        entity_matrix = self.encode(entity_docs)
        dim = entity_matrix.shape[1]

        entity_index = faiss.IndexFlatIP(dim)
        entity_index.add(entity_matrix)

        self.entity_store = {"entity_docs": entity_docs, "entity_ids": entity_ids, "entity_metas": entity_metas, "entity_index": entity_index}

    def build_relation_store(self, final_relations):
        relation_ids    = []
        relation_metas  = []
        relation_docs = []

        for i, rel in enumerate(final_relations):
            doc = (
                f"{rel['source']}. "
                f"{rel['target']}. "
                f"{rel['keywords']}."
                f"{rel['description']}"
            )
            relation_ids.append(f"rel_{i}")
            relation_metas.append(rel)
            relation_docs.append(doc)

        relation_matrix = self.encode(relation_docs)
        dim = relation_matrix.shape[1]
        relation_index  = faiss.IndexFlatIP(dim)
        relation_index.add(relation_matrix)

        self.relation_store = {"relation_ids": relation_ids, "relation_metas": relation_metas, "relation_index": relation_index}


    def query_entities_batch(self, ents, n=5):
        if self.entity_store is None:
            raise ValueError("Entity store not built")
        Q = self.encode(ents) # (N, d)
        scores, indices = self.entity_store["entity_index"].search(Q, n)
        results = []
        for qi in range(len(ents)):
            results.append([
                {
                    "score": round(float(scores[qi][i]), 4),
                    "doc": self.entity_store["entity_docs"][indices[qi][i]],
                    **self.entity_store["entity_metas"][indices[qi][i]]
                }
                for i in range(n)
                if indices[qi][i] != -1
            ])
        return results
````

## 1.11. Trích xuất keywords (seed entities, anchor entities) từ câu hỏi

````python
# Đúng ra phần này thuộc Retriever trong lý thuyết của báo cáo nhưng trong mã nguồn thì tôi để tách ra một lớp khác luôn
import re
import json

class SeedEntityExtractor:
    def __init__(self, llm: BaseLLM):
        self.llm = llm

    def _parse_response_data(self, text):
        try:
            # Tìm khối văn bản nằm giữa dấu ngoặc nhọn { ... }
            match = re.search(r'\{.*\}', text, re.DOTALL)
            if match:
                json_str = match.group()
                # parse chuỗi json
                result = json.loads(json_str)
                return result
            else:
                print("Không tìm thấy cấu trúc json trong phản hồi!")
                print(text)
                print()
                return None
        except json.JSONDecodeError as e:
            print(f'Lỗi định dạng JSON: {e}')
            return None

    def parse_batch(self, responses):
        batch_results = []
        for response in responses:
            batch_results.append(self._parse_response_data(response))
        return batch_results

    def extract_batch(self, questions: list[str]):
        system_prompt = PROMPTS["seed_entity_extraction_system_prompt"]

        user_prompts = []
        for question in questions:
            user_prompt = PROMPTS["seed_entity_extraction_user_prompt"].format(
                question = question
            )
            user_prompts.append(user_prompt)

        responses = self.llm.generate_batch(
            system_prompt,
            user_prompts,
            max_new_tokens=512,
            do_sample=False
        )
        return responses
````

## 1.12. Trích xuất thông tin để trả lời

````python
"""
Ở đây chứa các thành phần còn lại của module Retrieval - không bao gồn phần trích xuất thực thể hạt giống từ câu hỏi -> đã làm ở bước trên
Nhắc lại:

graph_store func:
    get_node(node_id) -> dict | None
    get_neighbors(node_id) -> list[{"neighbor": dict, "relation": dict}]

vector_store func:
    query_entities_batch([seed_text], n=top_k)

Optional graph_store func:
    find_nodes_by_name(name) -> list[dict]
    get_node_by_name(name) -> dict | None
"""

from __future__ import annotations

import time
from collections import defaultdict, deque
from collections.abc import Iterable
from typing import Any, Dict, Hashable, List, Optional, Set, Tuple

import networkx as nx
import numpy as np
from scipy.sparse import csr_matrix, lil_matrix

NodeId = Hashable
EdgeKey = Tuple[NodeId, NodeId]


def _ensure_string(value: Any) -> str:
    if isinstance(value, str):
        return value
    if isinstance(value, bytes):
        return value.decode("utf-8", errors="ignore")
    if isinstance(value, np.ndarray):
        return str(value[0]) if len(value) > 0 else ""
    if isinstance(value, np.generic):
        return str(value.item())
    return str(value)


def _ensure_list(value: Any) -> list:
    if value is None:
        return []
    if isinstance(value, list):
        return value
    if isinstance(value, np.ndarray):
        return value.tolist()
    if isinstance(value, (tuple, set)):
        return list(value)
    return [value]


def _ensure_int(value: Any) -> int:
    try:
        if isinstance(value, (int, np.integer)):
            return int(value)
        if isinstance(value, float):
            return int(value)
        if isinstance(value, np.ndarray):
            flat = value.flatten()
            return int(flat[0]) if flat.size > 0 else 0
        if isinstance(value, str):
            return int(value.strip())
        return int(value)
    except (ValueError, TypeError, IndexError):
        return 0


class Retriever:
    DEFAULT_CONFIG = {
        "top_k": 10,
        
        "seed_resolver_top_k": 5,
        "seed_resolver_threshold": 0.6,
        "seed_resolver_fallback_top_k": 3,
        "use_vector_seed_resolution": True,
        "use_vector_scores_as_seed_weights": True,
        "top_ppr_nodes": 100,
        "alpha": 0.8,
        "max_iter": 50,
        "convergence_eps": 1e-6,
        "resource_threshold": 1e-6,
        "max_component_size": 5000,
        "fallback_hop_radius": 2,
        "direct_isolated_score": 1.0,

        # isolated node neighbor fallback
        "use_isolated_neighbor_fallback": True,
        "isolated_neighbor_before": 2,
        "isolated_neighbor_after": 2,
        "isolated_neighbor_decay": 0.85,
        "isolated_neighbor_score_ratio": 0.8,

        "path_cutoff": 5,
        "top_paths": 20,
        "max_path_per_pair": 5,
        "path_search_timeout": 10.0,
        "use_paths": True,
        "weight_ppr_node": 0.8,
        "weight_path": 0.2,
        # rerank
        "candidate_multiplier": 5,
        "min_candidate_count": 30,
        "use_reranker": True,
        "rerank_weight": 0.75,
        "graph_weight_after_rerank": 0.25,
    }

    NAME_FIELDS = (
        "name",
        "canonical_name",
        "text",
        "entity",
        "alias",
        "aliases",
        "synonym",
        "synonyms",
    )

    def __init__(
        self,
        graph_store,
        chunks: List[Dict],
        vector_store=None,
        reranker=None,
        config: Dict = None,
    ):
        if isinstance(vector_store, dict) and config is None:
            config = vector_store
            vector_store = None
    
        if isinstance(reranker, dict) and config is None:
            config = reranker
            reranker = None
    
        self.graph_store = graph_store
        self.vector_store = vector_store
        self.reranker = reranker
        self.chunks = chunks or []
        self.config = {**self.DEFAULT_CONFIG, **(config or {})}
    
        self._idx_to_chunk = {
            _ensure_int(c["chunk_index"]): c
            for c in self.chunks
            if isinstance(c, dict) and "chunk_index" in c
        }
    
        self._name_to_node_ids: Optional[Dict[str, List[NodeId]]] = None

    def retrieve(
        self,
        seed_entities: List[Any],
        question: str = None,
        top_k: int = None,
    ) -> Dict[str, Any]:
        
        top_k = top_k or self.config["top_k"]
        seed_list = _ensure_list(seed_entities)
        seed_weights = self._resolve_seed_nodes(seed_list)

        if not seed_weights:
            return self._empty_result(
                input_seed_count=len(seed_list),
                warning="No seed could be resolved to graph nodes",
            )

        G, id_to_ent, id_tuple_to_rel, isolated_seed_weights = (
            self._build_seed_component_subgraph(seed_weights)
        )

        graph_scores: Dict[int, float] = {}

        # 1. Score chunk chứa isolated node
        isolated_scores = self._score_isolated_seed_chunks(isolated_seed_weights)
        for chunk_id, score in isolated_scores.items():
            graph_scores[chunk_id] = max(graph_scores.get(chunk_id, 0.0), score)
        
        # 2. lấy cac chunk liền trước / liền sau của chunk chứa isolated node
        isolated_neighbor_scores = self._score_isolated_seed_neighbor_chunks(
            isolated_seed_weights
        )
        for chunk_id, score in isolated_neighbor_scores.items():
            graph_scores[chunk_id] = max(graph_scores.get(chunk_id, 0.0), score)
            

        ppr_scores: Dict[NodeId, float] = {}
        active_seed_weights = {
            node_id: weight
            for node_id, weight in seed_weights.items()
            if node_id in G
        }

        if G.number_of_nodes() > 0 and active_seed_weights:
            ppr_scores = self._run_ppr(G, active_seed_weights)

            ppr_chunk_scores = self._score_chunks_by_ppr_nodes(
                ppr_scores,
                id_to_ent,
                id_tuple_to_rel,
                top_nodes=self.config["top_ppr_nodes"],
            )

            path_chunk_scores: Dict[int, float] = {}
            if self.config.get("use_paths", True):
                paths = self._find_paths(G, ppr_scores, list(active_seed_weights.keys()))
                path_chunk_scores = self._score_chunks_by_weighted_paths(
                    paths,
                    id_to_ent,
                    id_tuple_to_rel,
                )

            for chunk_id in set(ppr_chunk_scores) | set(path_chunk_scores):
                score = (
                    self.config["weight_ppr_node"]
                    * ppr_chunk_scores.get(chunk_id, 0.0)
                    + self.config["weight_path"]
                    * path_chunk_scores.get(chunk_id, 0.0)
                )
                graph_scores[chunk_id] = max(graph_scores.get(chunk_id, 0.0), score)

        sorted_pairs = sorted(graph_scores.items(), key=lambda x: x[1], reverse=True)

        candidate_count = max(
            top_k * self.config.get("candidate_multiplier", 5),
            self.config.get("min_candidate_count", 30),
        )
        
        candidate_pairs = [
            (chunk_id, score)
            for chunk_id, score in sorted_pairs
            if chunk_id in self._idx_to_chunk
        ][:candidate_count]
        
        candidate_chunks = []
        for chunk_id, score in candidate_pairs:
            chunk = self._idx_to_chunk.get(chunk_id)
            if chunk is not None:
                candidate_chunks.append({
                    **chunk,
                    "graph_score": float(score),
                })
                
        rerank_query = question or " ".join(_ensure_string(x) for x in seed_list)

        if self.config.get("use_reranker", True) and self.reranker is not None:
            final_chunks = self._rerank_chunks(
                query=rerank_query,
                chunks=candidate_chunks,
                top_k=top_k,
            )
        else:
            final_chunks = candidate_chunks[:top_k]
        
        top_pairs = [
            (
                _ensure_int(chunk.get("chunk_index")),
                float(chunk.get("final_score", chunk.get("graph_score", 0.0))),
            )
            for chunk in final_chunks
        ]

        return {
            "final_chunk_ids": [chunk_id for chunk_id, _ in top_pairs],
            "final_chunks": final_chunks,
            "graph_scores": {
                chunk_id: float(score)
                for chunk_id, score in top_pairs
            },
            "retrieval_details": {
                "input_seed_count": len(seed_list),
                "resolved_seed_count": len(seed_weights),
                "isolated_seed_count": len(isolated_seed_weights),
                "subgraph_node_count": G.number_of_nodes(),
                "subgraph_edge_count": G.number_of_edges(),
                "ppr_node_count": len(ppr_scores),
                "candidate_count_before_rerank": len(candidate_chunks),
                "isolated_direct_chunk_count": len(isolated_scores),
                "isolated_neighbor_chunk_count": len(isolated_neighbor_scores),
                "used_isolated_neighbor_fallback": bool(
                    self.config.get("use_isolated_neighbor_fallback", True)
                    and len(isolated_neighbor_scores) > 0
                ),
                "used_reranker": bool(
                    self.config.get("use_reranker", True)
                    and self.reranker is not None
                ),
                "final_count": len(final_chunks),
            },
        }

    def _rerank_chunks(
        self,
        query: str,
        chunks: List[Dict],
        top_k: int,
    ) -> List[Dict]:
        if not chunks:
            return []
    
        if self.reranker is None:
            return chunks[:top_k]
    
        texts = [
            _ensure_string(chunk.get("text", ""))
            for chunk in chunks
        ]
    
        try:
            rerank_scores = self.reranker.score(query, texts)
        except Exception as exc:
            print(f"[WARNING] Reranking failed: {exc}")
            return chunks[:top_k]
    
        if not rerank_scores or len(rerank_scores) != len(chunks):
            print(
                f"[WARNING] Invalid rerank scores: "
                f"{len(rerank_scores) if rerank_scores else 0} scores for {len(chunks)} chunks"
            )
            return chunks[:top_k]
    
        rerank_scores_norm = self._minmax_normalize_list(rerank_scores)
    
        graph_scores = [
            float(chunk.get("graph_score", 0.0))
            for chunk in chunks
        ]
        graph_scores_norm = self._minmax_normalize_list(graph_scores)
    
        rerank_weight = float(self.config.get("rerank_weight", 0.75))
        graph_weight = float(self.config.get("graph_weight_after_rerank", 0.25))
    
        scored_chunks = []
    
        for chunk, rerank_score, rerank_norm, graph_score, graph_norm in zip(
            chunks,
            rerank_scores,
            rerank_scores_norm,
            graph_scores,
            graph_scores_norm,
        ):
            final_score = rerank_weight * rerank_norm + graph_weight * graph_norm
    
            scored_chunks.append({
                **chunk,
                "rerank_score": float(rerank_score),
                "rerank_score_norm": float(rerank_norm),
                "graph_score_norm": float(graph_norm),
                "final_score": float(final_score),
            })
    
        scored_chunks.sort(key=lambda x: x["final_score"], reverse=True)
    
        return scored_chunks[:top_k]

    def _minmax_normalize_list(self, values: List[float]) -> List[float]:
        if not values:
            return []
    
        values = [float(v) for v in values]
        min_v = min(values)
        max_v = max(values)
    
        if max_v - min_v < 1e-12:
            return [1.0 for _ in values]
    
        return [
            (v - min_v) / (max_v - min_v)
            for v in values
        ]
        
    def _resolve_seed_nodes(self, seed_entities: List[Any]) -> Dict[NodeId, float]:
        seed_weights: Dict[NodeId, float] = {}

        for seed in _ensure_list(seed_entities):
            direct_node_id = self._resolve_direct_node_id(seed)
            if direct_node_id is not None:
                if self.graph_store.get_node(direct_node_id) is not None:
                    seed_weights[direct_node_id] = max(
                        seed_weights.get(direct_node_id, 0.0),
                        1.0,
                    )
                continue

            seed_text = _ensure_string(seed).strip()
            if not seed_text:
                continue

            for node in self._find_graph_nodes_by_name(seed_text):
                matched_id = self._node_id_from_node(node)
                if matched_id is None:
                    continue
                if self.graph_store.get_node(matched_id) is not None:
                    seed_weights[matched_id] = max(
                        seed_weights.get(matched_id, 0.0),
                        1.0,
                    )

            for matched_id, weight in self._find_graph_nodes_by_vector(seed_text):
                seed_weights[matched_id] = max(
                    seed_weights.get(matched_id, 0.0),
                    weight,
                )

        return seed_weights
        
    def _find_graph_nodes_by_vector(self, seed_text: str) -> List[Tuple[NodeId, float]]:
        if not self.config.get("use_vector_seed_resolution", True):
            return []
    
        if self.vector_store is None or not hasattr(self.vector_store, "query_entities_batch"):
            return []
    
        top_k = self.config["seed_resolver_top_k"]
        threshold = self.config["seed_resolver_threshold"]
        use_scores = self.config.get("use_vector_scores_as_seed_weights", True)
    
        # Số lượng fallback hay vẫn cố gắng giữ lại nếu tất cả score < threshold
        fallback_top_k = self.config.get("seed_resolver_fallback_top_k", top_k)
    
        try:
            batch_results = self.vector_store.query_entities_batch([seed_text], n=top_k) or []
        except Exception as exc:
            print(f"[WARNING] Vector seed resolution failed for '{seed_text}': {exc}")
            return []
    
        if not batch_results:
            return []
    
        result_items = batch_results[0] if isinstance(batch_results[0], list) else batch_results
    
        candidates: List[Tuple[NodeId, float]] = []
    
        for item in _ensure_list(result_items):
            if not isinstance(item, dict):
                continue
    
            score = self._vector_score(item)
    
            raw_node_id = self._node_id_from_vector_result(item)
            graph_node_id = self._resolve_graph_node_id(raw_node_id)
    
            if graph_node_id is None:
                continue
    
            candidates.append((graph_node_id, float(score)))
    
        if not candidates:
            return []
    
        # Sắp xếp giảm dần theo score
        candidates = sorted(candidates, key=lambda x: x[1], reverse=True)
    
        # TH1: Có entity đạt threshold
        above_threshold = [
            (node_id, score)
            for node_id, score in candidates
            if score >= threshold
        ]
    
        if above_threshold:
            selected = above_threshold
        else:
            # TH2: Tất cả đều dưới threshold -> lấy top fallback
            selected = candidates[:fallback_top_k]
    
        matches: List[Tuple[NodeId, float]] = []
    
        for graph_node_id, score in selected:
            weight = score if use_scores and score > 0 else 1.0
            matches.append((graph_node_id, float(weight)))
    
        return matches
    
    def _find_graph_nodes_by_name(self, name: str) -> List[Dict]:
        norm = self._normalize_name(name)
        if not norm:
            return []

        if hasattr(self.graph_store, "find_nodes_by_name"):
            return self.graph_store.find_nodes_by_name(name) or []

        if hasattr(self.graph_store, "get_node_by_name"):
            node = self.graph_store.get_node_by_name(name)
            return [node] if node else []

        if self._name_to_node_ids is None:
            self._build_name_index_if_possible()

        node_ids = (self._name_to_node_ids or {}).get(norm, [])
        nodes = []
        for node_id in node_ids:
            node = self.graph_store.get_node(node_id)
            if node is not None:
                nodes.append(node)
        return nodes

    def _build_seed_component_subgraph(
        self,
        seed_weights: Dict[NodeId, float],
    ) -> Tuple[nx.Graph, Dict[NodeId, Dict], Dict[EdgeKey, Dict], Dict[NodeId, float]]:
        G = nx.Graph()
        id_to_ent: Dict[NodeId, Dict] = {}
        id_tuple_to_rel: Dict[EdgeKey, Dict] = {}
        isolated_seed_weights: Dict[NodeId, float] = {}
        processed_component_nodes: Set[NodeId] = set()

        for seed_id, seed_weight in seed_weights.items():
            if seed_id in processed_component_nodes:
                continue

            component_nodes, component_edges, component_ents, reached_limit = (
                self._bfs_component_from_seed(
                    seed_id,
                    max_nodes=self.config["max_component_size"],
                )
            )
            if not component_nodes:
                continue
            if len(component_nodes) <= 1 and not component_edges:
                isolated_seed_weights[seed_id] = seed_weight
                processed_component_nodes.update(component_nodes)
                continue

            if reached_limit:
                component_nodes, component_edges, component_ents = (
                    self._bfs_ego_graph_from_seed(
                        seed_id,
                        radius=self.config["fallback_hop_radius"],
                    )
                )

            processed_component_nodes.update(component_nodes)
            id_to_ent.update(component_ents)
            id_tuple_to_rel.update(component_edges)

            G.add_nodes_from(component_nodes)
            for (src, tgt), relation in component_edges.items():
                if src in component_nodes and tgt in component_nodes:
                    G.add_edge(src, tgt, **(relation or {}))

        return G, id_to_ent, id_tuple_to_rel, isolated_seed_weights

    def _bfs_component_from_seed(
        self,
        seed_id: NodeId,
        max_nodes: int,
    ) -> Tuple[Set[NodeId], Dict[EdgeKey, Dict], Dict[NodeId, Dict], bool]:
        seed_ent = self.graph_store.get_node(seed_id)
        if seed_ent is None:
            return set(), {}, {}, False

        visited: Set[NodeId] = {seed_id}
        queue = deque([seed_id])
        edges: Dict[EdgeKey, Dict] = {}
        id_to_ent: Dict[NodeId, Dict] = {seed_id: seed_ent}

        while queue:
            current_id = queue.popleft()
            for nb in self._safe_get_neighbors(current_id):
                neighbor = nb.get("neighbor") or {}
                relation = nb.get("relation") or {}
                neighbor_id = self._neighbor_id(current_id, neighbor, relation)
                if neighbor_id is None:
                    continue

                id_to_ent.setdefault(neighbor_id, neighbor)
                edge_key = self._relation_edge_key(current_id, neighbor_id, relation)
                if edge_key is not None:
                    edges[edge_key] = relation

                if neighbor_id in visited:
                    continue

                visited.add(neighbor_id)
                if len(visited) > max_nodes:
                    return visited, edges, id_to_ent, True
                queue.append(neighbor_id)

        return visited, edges, id_to_ent, False

    def _bfs_ego_graph_from_seed(
        self,
        seed_id: NodeId,
        radius: int,
    ) -> Tuple[Set[NodeId], Dict[EdgeKey, Dict], Dict[NodeId, Dict]]:
        seed_ent = self.graph_store.get_node(seed_id)
        if seed_ent is None:
            return set(), {}, {}

        visited: Set[NodeId] = {seed_id}
        queue = deque([(seed_id, 0)])
        edges: Dict[EdgeKey, Dict] = {}
        id_to_ent: Dict[NodeId, Dict] = {seed_id: seed_ent}

        while queue:
            current_id, depth = queue.popleft()
            if depth >= radius:
                continue

            for nb in self._safe_get_neighbors(current_id):
                neighbor = nb.get("neighbor") or {}
                relation = nb.get("relation") or {}
                neighbor_id = self._neighbor_id(current_id, neighbor, relation)
                if neighbor_id is None:
                    continue

                id_to_ent.setdefault(neighbor_id, neighbor)
                edge_key = self._relation_edge_key(current_id, neighbor_id, relation)
                if edge_key is not None:
                    edges[edge_key] = relation

                if neighbor_id in visited:
                    continue

                visited.add(neighbor_id)
                queue.append((neighbor_id, depth + 1))

        return visited, edges, id_to_ent

    def _score_isolated_seed_chunks(
        self,
        isolated_seed_weights: Dict[NodeId, float],
    ) -> Dict[int, float]:
        if not isolated_seed_weights:
            return {}
    
        max_weight = max(isolated_seed_weights.values()) or 1.0
        chunk_scores: Dict[int, float] = {}
    
        for node_id, seed_weight in isolated_seed_weights.items():
            ent = self.graph_store.get_node(node_id)
            if ent is None:
                continue
    
            score = (
                seed_weight
                / max(max_weight, 1e-9)
                * self.config["direct_isolated_score"]
            )
    
            for chunk_id in self._chunk_ids_from_value(ent.get("chunk_index")):
                chunk_scores[chunk_id] = max(chunk_scores.get(chunk_id, 0.0), score)
    
        return chunk_scores

    def _score_isolated_seed_neighbor_chunks(
        self,
        isolated_seed_weights: Dict[NodeId, float],
    ) -> Dict[int, float]:
        """
        Fallback cho từng isolated seed node hay seed node bị cô lập.
    
        Với mỗi isolated node:
            - lấy chunk chứa node đó làm anchor
            - lấy thêm N chunk liền trước và M chunk liền sau (N, M định trước trong input)
            - score giảm theo khoảng cách
    
        Ví dụ:
            anchor chunk = 10
            before = 2, after = 2
            => lấy chunk 8, 9, 10, 11, 12
        """
        if not isolated_seed_weights:
            return {}

        if not self.config.get("use_isolated_neighbor_fallback", True):
            return {}
    
        before = int(self.config.get("isolated_neighbor_before", 2))
        after = int(self.config.get("isolated_neighbor_after", 2))
        decay = float(self.config.get("isolated_neighbor_decay", 0.85))
        score_ratio = float(self.config.get("isolated_neighbor_score_ratio", 0.8))
    
        max_weight = max(isolated_seed_weights.values()) or 1.0
        chunk_scores: Dict[int, float] = {}
    
        for node_id, seed_weight in isolated_seed_weights.items():
            ent = self.graph_store.get_node(node_id)
            if ent is None:
                continue
    
            base_score = (
                seed_weight
                / max(max_weight, 1e-9)
                * self.config["direct_isolated_score"]
            )
    
            anchor_chunk_ids = self._chunk_ids_from_value(ent.get("chunk_index"))
    
            for anchor_chunk_id in anchor_chunk_ids:
                anchor_chunk_id = _ensure_int(anchor_chunk_id)
                anchor_chunk = self._idx_to_chunk.get(anchor_chunk_id)
    
                if anchor_chunk is None:
                    continue
    
                anchor_doc_id = (
                    anchor_chunk.get("doc_id")
                    or anchor_chunk.get("source_id")
                    or anchor_chunk.get("file_id")
                )
    
                for offset in range(-before, after + 1):
                    neighbor_chunk_id = anchor_chunk_id + offset
                    neighbor_chunk = self._idx_to_chunk.get(neighbor_chunk_id)
    
                    if neighbor_chunk is None:
                        continue
    
                    neighbor_doc_id = (
                        neighbor_chunk.get("doc_id")
                        or neighbor_chunk.get("source_id")
                        or neighbor_chunk.get("file_id")
                    )
    
                    if anchor_doc_id is not None and neighbor_doc_id != anchor_doc_id:
                        continue
    
                    distance = abs(offset)
    
                    if distance == 0:
                        score = base_score
                    else:
                        score = base_score * score_ratio * (decay ** distance)
    
                    chunk_scores[neighbor_chunk_id] = max(
                        chunk_scores.get(neighbor_chunk_id, 0.0),
                        float(score),
                    )
    
        return chunk_scores

    def _run_ppr(self, G: nx.Graph, seed_weights: Dict[NodeId, float]) -> Dict[NodeId, float]:
        nodes = list(G.nodes)
        if not nodes:
            return {}

        node_idx = {node_id: idx for idx, node_id in enumerate(nodes)}
        valid_seed_weights = {
            node_id: weight
            for node_id, weight in seed_weights.items()
            if node_id in node_idx and weight > 0
        }
        total_seed_weight = sum(valid_seed_weights.values())
        if total_seed_weight <= 0:
            return {}

        n = len(nodes)
        e = np.zeros(n, dtype=float)
        for seed_id, seed_weight in valid_seed_weights.items():
            e[node_idx[seed_id]] = seed_weight / total_seed_weight

        transition = lil_matrix((n, n), dtype=float)
        for current in nodes:
            current_idx = node_idx[current]
            neighbors = list(G.neighbors(current))

            if neighbors:
                probability = 1.0 / len(neighbors)
                for neighbor in neighbors:
                    transition[node_idx[neighbor], current_idx] = probability
            else:
                for restart_idx, restart_probability in enumerate(e):
                    if restart_probability > 0:
                        transition[restart_idx, current_idx] = restart_probability

        transition = csr_matrix(transition)
        scores = e.copy()
        alpha = self.config["alpha"]

        for _ in range(self.config["max_iter"]):
            next_scores = alpha * transition.dot(scores) + (1.0 - alpha) * e
            total = float(next_scores.sum())
            if total > 0:
                next_scores = next_scores / total

            if np.linalg.norm(next_scores - scores, 1) < self.config["convergence_eps"]:
                scores = next_scores
                break

            scores = next_scores

        return {
            node_id: float(scores[node_idx[node_id]])
            for node_id in nodes
        }

    def _score_chunks_by_ppr_nodes(
        self,
        S: Dict[NodeId, float],
        id_to_ent: Dict[NodeId, Dict],
        id_tuple_to_rel: Dict[EdgeKey, Dict],
        top_nodes: int = 100,
    ) -> Dict[int, float]:
        selected_nodes = sorted(S.items(), key=lambda x: x[1], reverse=True)[:top_nodes]
        selected_node_ids = {node_id for node_id, _ in selected_nodes}
        chunk_scores: Dict[int, float] = defaultdict(float)

        for node_id, node_score in selected_nodes:
            ent = id_to_ent.get(node_id) or self.graph_store.get_node(node_id)
            if not ent:
                continue
            for chunk_id in self._chunk_ids_from_value(ent.get("chunk_index")):
                chunk_scores[chunk_id] += float(node_score)

        for (src, tgt), relation in id_tuple_to_rel.items():
            if src not in selected_node_ids and tgt not in selected_node_ids:
                continue
            relation_score = (S.get(src, 0.0) + S.get(tgt, 0.0)) / 2.0
            for chunk_id in self._chunk_ids_from_value(
                (relation or {}).get("chunk_index")
            ):
                chunk_scores[chunk_id] += float(relation_score)

        return self._normalize_scores(chunk_scores)

    def _find_paths(
        self,
        G: nx.Graph,
        S: Dict[NodeId, float],
        seed_nodes: List[NodeId],
    ) -> List[Dict]:
        start_nodes = [node_id for node_id in seed_nodes if node_id in G]
        if not start_nodes or not S:
            return []

        end_candidates = [
            node_id
            for node_id, score in sorted(S.items(), key=lambda x: x[1], reverse=True)
            if node_id in G and score >= self.config["resource_threshold"]
        ][: self.config["top_ppr_nodes"]]

        deadline = time.monotonic() + self.config["path_search_timeout"]
        paths: List[Dict] = []
        seen_paths: Set[Tuple[NodeId, ...]] = set()

        for start in start_nodes:
            for end in end_candidates:
                if time.monotonic() >= deadline:
                    break
                if start == end:
                    continue

                try:
                    path_generator = nx.shortest_simple_paths(G, start, end)
                    path_count = 0

                    for path in path_generator:
                        if time.monotonic() >= deadline:
                            break
                        if len(path) - 1 > self.config["path_cutoff"]:
                            break

                        path_key = tuple(path)
                        if path_key in seen_paths:
                            continue

                        seen_paths.add(path_key)
                        path_score = sum(S.get(node_id, 0.0) for node_id in path)
                        path_score = path_score / max(1, len(path) - 1)

                        paths.append({
                            "nodes": path,
                            "relations": list(zip(path[:-1], path[1:])),
                            "score": float(path_score),
                        })

                        path_count += 1
                        if path_count >= self.config["max_path_per_pair"]:
                            break

                except (nx.NetworkXNoPath, nx.NodeNotFound):
                    continue

            if time.monotonic() >= deadline:
                break

        paths.sort(key=lambda x: x["score"], reverse=True)
        return paths[: self.config["top_paths"]]

    def _score_chunks_by_weighted_paths(
        self,
        paths: List[Dict],
        id_to_ent: Dict[NodeId, Dict],
        id_tuple_to_rel: Dict[EdgeKey, Dict],
    ) -> Dict[int, float]:
        chunk_scores: Dict[int, float] = defaultdict(float)

        for path in paths:
            path_score = float(path.get("score", 0.0))

            for node_id in path.get("nodes", []):
                ent = id_to_ent.get(node_id) or self.graph_store.get_node(node_id)
                if not ent:
                    continue
                for chunk_id in self._chunk_ids_from_value(ent.get("chunk_index")):
                    chunk_scores[chunk_id] += path_score

            for src, tgt in path.get("relations", []):
                relation = id_tuple_to_rel.get((src, tgt)) or id_tuple_to_rel.get((tgt, src))
                if not relation:
                    continue
                for chunk_id in self._chunk_ids_from_value(relation.get("chunk_index")):
                    chunk_scores[chunk_id] += path_score

        return self._normalize_scores(chunk_scores)

    def _build_name_index_if_possible(self) -> None:
        self._name_to_node_ids = defaultdict(list)

        graph = getattr(self.graph_store, "graph", None)
        if graph is not None and hasattr(graph, "nodes"):
            for raw_node_id in graph.nodes:
                node_id = raw_node_id
                node = self.graph_store.get_node(node_id)
                if node:
                    self._index_node_names(node_id, node)

                graph_node_data = graph.nodes[raw_node_id]
                metas = graph_node_data.get("metas", {}) if isinstance(graph_node_data, dict) else {}
                raw_self = metas.get("self", {}) if isinstance(metas, dict) else {}
                if isinstance(raw_self, dict):
                    self._index_node_names(node_id, raw_self)
            return

        iterable_nodes = getattr(self.graph_store, "nodes", None)
        if iterable_nodes is None:
            return

        for node in iterable_nodes:
            if not isinstance(node, dict):
                continue
            node_id = self._node_id_from_node(node)
            if node_id is not None:
                self._index_node_names(node_id, node)

    def _index_node_names(self, node_id: NodeId, node: Dict) -> None:
        for value in self._candidate_name_values(node):
            norm = self._normalize_name(value)
            if norm and node_id not in self._name_to_node_ids[norm]:
                self._name_to_node_ids[norm].append(node_id)

    def _candidate_name_values(self, node: Dict) -> List[str]:
        values = []
        for field in self.NAME_FIELDS:
            value = node.get(field)
            if value is None:
                continue
            if isinstance(value, (list, tuple, set, np.ndarray)):
                values.extend(_ensure_string(item) for item in value if item is not None)
            else:
                values.append(_ensure_string(value))
        return values

    def _resolve_direct_node_id(self, seed: Any) -> Optional[NodeId]:
        if isinstance(seed, (int, np.integer)):
            return self._resolve_graph_node_id(int(seed))

        if isinstance(seed, str):
            seed_text = seed.strip()
            if not seed_text:
                return None
            return self._resolve_graph_node_id(seed_text)

        return None

    def _resolve_graph_node_id(self, raw_node_id: Any) -> Optional[NodeId]:
        if raw_node_id is None:
            return None

        if isinstance(raw_node_id, np.generic):
            raw_node_id = raw_node_id.item()

        candidates = [raw_node_id]

        if not isinstance(raw_node_id, str):
            candidates.append(str(raw_node_id))
        else:
            stripped = raw_node_id.strip()
            if stripped != raw_node_id:
                candidates.append(stripped)
            try:
                candidates.append(int(stripped))
            except ValueError:
                pass

        seen = set()
        for candidate in candidates:
            try:
                marker = (type(candidate), candidate)
                if marker in seen:
                    continue
                seen.add(marker)
            except TypeError:
                pass

            if self.graph_store.get_node(candidate) is not None:
                return candidate

        return None

    def _node_id_from_vector_result(self, item: Dict) -> Optional[NodeId]:
        self_data = item.get("self")
        if isinstance(self_data, dict):
            for field in ("id", "node_id", "entity_id"):
                if self_data.get(field) is not None:
                    return self_data[field]

        for field in ("id", "node_id", "entity_id"):
            if item.get(field) is not None:
                return item[field]

        return None

    def _vector_score(self, item: Dict) -> float:
        raw_score = item.get("score", 1.0)
        try:
            if isinstance(raw_score, np.ndarray):
                flat = raw_score.flatten()
                return float(flat[0]) if flat.size > 0 else 0.0
            if isinstance(raw_score, np.generic):
                return float(raw_score.item())
            return float(raw_score)
        except (TypeError, ValueError):
            return 1.0

    def _node_id_from_node(self, node: Dict) -> Optional[NodeId]:
        if not isinstance(node, dict):
            return None

        if node.get("id") is not None:
            return node["id"]

        self_data = node.get("self")
        if isinstance(self_data, dict) and self_data.get("id") is not None:
            return self_data["id"]

        return None

    def _safe_get_neighbors(self, node_id: NodeId) -> List[Dict]:
        try:
            return self.graph_store.get_neighbors(node_id) or []
        except Exception:
            return []

    def _neighbor_id(
        self,
        current_id: NodeId,
        neighbor: Dict,
        relation: Dict,
    ) -> Optional[NodeId]:
        if isinstance(neighbor, dict) and neighbor.get("id") is not None:
            return neighbor["id"]

        src = relation.get("source_id")
        tgt = relation.get("target_id")
        if src is None or tgt is None:
            return None

        if self._ids_equal(src, current_id):
            return tgt
        if self._ids_equal(tgt, current_id):
            return src
        return None

    def _relation_edge_key(
        self,
        current_id: NodeId,
        neighbor_id: NodeId,
        relation: Dict,
    ) -> Optional[EdgeKey]:
        src = relation.get("source_id")
        tgt = relation.get("target_id")
        if src is not None and tgt is not None:
            if self._ids_equal(src, current_id) and self._ids_equal(tgt, neighbor_id):
                return current_id, neighbor_id
            if self._ids_equal(src, neighbor_id) and self._ids_equal(tgt, current_id):
                return neighbor_id, current_id
        return current_id, neighbor_id

    def _ids_equal(self, left: Any, right: Any) -> bool:
        return left == right or _ensure_string(left) == _ensure_string(right)

    def _chunk_ids_from_value(self, value: Any) -> List[int]:
        if value is None:
            return []

        if isinstance(value, np.ndarray):
            return [
                _ensure_int(item)
                for item in value.flatten().tolist()
                if item is not None
            ]

        if isinstance(value, Iterable) and not isinstance(value, (str, bytes, dict)):
            return [_ensure_int(item) for item in value if item is not None]

        return [_ensure_int(value)]

    def _normalize_name(self, name: str) -> str:
        return _ensure_string(name).strip().lower()

    def _normalize_scores(self, scores: Dict[int, float]) -> Dict[int, float]:
        if not scores:
            return {}

        max_score = max(float(score) for score in scores.values())
        if max_score <= 0:
            return {chunk_id: 0.0 for chunk_id in scores}

        return {
            chunk_id: float(score) / max_score
            for chunk_id, score in scores.items()
        }

    def _empty_result(self, input_seed_count: int, warning: str = None) -> Dict[str, Any]:
        details = {
            "input_seed_count": input_seed_count,
            "resolved_seed_count": 0,
            "isolated_seed_count": 0,
            "subgraph_node_count": 0,
            "subgraph_edge_count": 0,
            "ppr_node_count": 0,
            "final_count": 0,
        }
        if warning:
            details["warning"] = warning

        return {
            "final_chunk_ids": [],
            "final_chunks": [],
            "graph_scores": {},
            "retrieval_details": details,
        }
````

## 1.13. Trả lời câu hỏi

````python
import re
import json
from typing import Dict, Any

# Cái này dùng trong hệ thống với model local như ở các bước trước, 
# tuy nhiên, để đánh giá thì phải lấy đầu ra của bước retrieval, chuyển thành đầu vào thích hợp để sinh câu trả lời (nhằm đánh giá), 
# sau đó đánh giá thông qua model gpt4o mini bằng cách sử dụng thư mục Generation - chi tiết làm như nào có ghi trong đó 
class AnswerGenerator:
    def __init__(self, llm: BaseLLM, max_context_tokens=8000):
        self.llm = llm
        self.max_context_tokens = max_context_tokens

    def _truncate_text_by_tokens(self, text: str, max_tokens: int) -> str:
        tokenizer = self.llm.tokenizer

        token_ids = tokenizer.encode(
            text,
            add_special_tokens=False
        )

        if len(token_ids) <= max_tokens:
            return text

        token_ids = token_ids[:max_tokens]

        return tokenizer.decode(
            token_ids,
            skip_special_tokens=True
        )
        
    def answer_batch(self, contexts):
        system_prompt = PROMPTS["question_answering_system_prompt"]
        user_prompts = []
        effective_contexts = []
    
        for context in contexts:
            chunk_context = context.get("chunk_context", "")
    
            if isinstance(chunk_context, list):
                chunk_context_text = "\n".join(str(chunk) for chunk in chunk_context)
            else:
                chunk_context_text = str(chunk_context)
    
            truncated_context = self._truncate_text_by_tokens(
                chunk_context_text,
                self.max_context_tokens
            )
    
            # Lưu context thực sự đưa vào prompt
            effective_contexts.append([truncated_context])
    
            user_prompt = PROMPTS["question_answering_user_prompt"].format(
                question=context["question"],
                chunk_context=truncated_context
            )
            user_prompts.append(user_prompt)
    
        responses = self.llm.generate_batch(
            system_prompt,
            user_prompts,
            max_new_tokens=512,
            do_sample=False
        )
    
        return responses, effective_contexts

    
    @staticmethod
    def _parse_qa_response(raw_text: str) -> Dict[str, str]:
        """
        Parse output của LLM dạng JSON:
        {
            "answer": "...",
            "reason": "..."
        }

        Có xử lý cả trường hợp model trả về ```json ... ```
        """

        if raw_text is None:
            return {
                "answer": "Not enough information",
                "reason": "Empty response from model."
            }

        text = raw_text.strip()

        # Bỏ markdown nếu có
        text = re.sub(r"^```json\s*", "", text, flags=re.IGNORECASE)
        text = re.sub(r"^```\s*", "", text)
        text = re.sub(r"\s*```$", "", text)
        text = text.strip()

        try:
            data = json.loads(text)
        except json.JSONDecodeError:
            lowered = text.lower()
        
            if lowered in ["i don't know", "unknown", "not enough information"]:
                return {
                    "answer": "Not enough information",
                    "reason": "The model indicated that the Knowledge Base does not contain enough information."
                }
        
            return {
                "answer": "Not enough information",
                "reason": f"Invalid JSON response: {raw_text[:200]}"
            }

        answer = data.get("answer", "")
        reason = data.get("reason", "")

        if not isinstance(answer, str):
            answer = str(answer)

        if not isinstance(reason, str):
            reason = str(reason)

        answer = answer.strip()
        reason = reason.strip()

        if answer == "":
            answer = "Not enough information"

        return {
            "answer": answer,
            "reason": reason
        }
    def parse_batch(self, responses):
        exact_answers = []
        for response in responses:
            answer = AnswerGenerator._parse_qa_response(response)
            exact_answers.append(answer)
        return exact_answers
````

## 1.14. Lớp RAG

````python
from dotenv import load_dotenv
from datasets import Dataset, load_dataset
import numpy as np
from collections import defaultdict
import os
import gc
import psutil
import torch

import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq

from tqdm import tqdm
import json
import copy

import traceback
from sentence_transformers import SentenceTransformer


from nltk.tokenize.punkt import PunktSentenceTokenizer, PunktParameters
punkt_param = PunktParameters()

punkt_param.abbrev_types = set(['Dr', 'Ms', 'Mr', 'Mrs', 'Prof', 'Inc', 'i.e', 'al', 'St', 'Ltd', 'Jr', 'Sr', 'Co', '.com', '.net', '.org', '.io', '.gov'])

sentence_splitter = PunktSentenceTokenizer(punkt_param)

class RAG:
    def __init__(self, corpus_path: str, llm=None, embedding_model=None, reranker=None):
        self.corpus_path = corpus_path
        self.llm = llm
        self.embedding_model = embedding_model
        self.reranker = reranker

        self.data_processor = DataProcessor(
            tokenizer=self.llm.tokenizer if self.llm else None,
            sentence_splitter=sentence_splitter
        )
        self.validator    = Validator()
        self.deduplicator = Deduplicator(embedding_model=self.embedding_model)
        self.graph_store     = GraphStore()
        self.vector_store    = VectorStore(embedding_model=self.embedding_model)


        self.retriever = Retriever(
            graph_store=self.graph_store,
            vector_store=self.vector_store,
            chunks=[],
            reranker=self.reranker,
            config={
                "top_k": 10,
            
                # Seed node dạng text -> graph node
                "seed_resolver_top_k": 5,
                "seed_resolver_threshold": 0.65,
                "seed_resolver_fallback_top_k": 1,
                "use_vector_seed_resolution": True,
                "use_vector_scores_as_seed_weights": True,
            
                # PPR: giới hạn lan truyền để bớt nhiễu
                "top_ppr_nodes": 30,
                "alpha": 0.6,
                "max_iter": 50,
                "convergence_eps": 1e-6,
            
                # Component/subgraph
                "resource_threshold": 1e-4,
                "max_component_size": 800,
                "fallback_hop_radius": 1,
            
                # Isolated node direct score
                "direct_isolated_score": 1.0,
            
                # Isolated node neighbor fallback
                # Nếu seed node bị isolated:
                #   lấy chunk chứa node + N chunk trước + M chunk sau
                "use_isolated_neighbor_fallback": True,
                "isolated_neighbor_before": 2,
                "isolated_neighbor_after": 2,
                "isolated_neighbor_decay": 0.85,
                "isolated_neighbor_score_ratio": 0.8,
            
                # Path: hiện tại không sử dụng
                "use_paths": False,
                "path_cutoff": 5,
                "top_paths": 20,
                "max_path_per_pair": 5,
                "path_search_timeout": 10.0,
            
                # Graph score
                "weight_ppr_node": 1.0,
                "weight_path": 0.0,
            
            
                # Reranker
                "use_reranker": True,
                "candidate_multiplier": 2,
                "min_candidate_count": 30,
                "rerank_weight": 0.6,
                "graph_weight_after_rerank": 0.4,
            },
        )

        self._extractor             = None
        self._seed_entity_extractor = None
        self._answer_generator      = None
        self._qa_judger             = None


    @property
    def extractor(self):
        self._require_llm("extractor")
        if self._extractor is None:
            self._extractor = Extractor(llm=self.llm)
        return self._extractor

    @property
    def seed_entity_extractor(self):
        self._require_llm("seed_entity_extractor")
        if self._seed_entity_extractor is None:
            self._seed_entity_extractor = SeedEntityExtractor(llm=self.llm)
        return self._seed_entity_extractor

    @property
    def answer_generator(self):
        self._require_llm("answer_generator")
        if self._answer_generator is None:
            self._answer_generator = AnswerGenerator(
                llm=self.llm,
                max_context_tokens=9000
            )
        return self._answer_generator

    @property
    def qa_judger(self):
        self._require_llm("qa_judger")
        if self._qa_judger is None:
            self._qa_judger = QA_Judger(llm=self.llm)
        return self._qa_judger

    def _require_llm(self, name: str):
        if self.llm is None:
            raise RuntimeError(
                f"'{name}' cần LLM nhưng llm=None. "
                f"Truyền llm khi khởi tạo RAG."
            )


    def doc_chunking(self, chunk_output_path, chunk_size=256):
        if os.path.exists(chunk_output_path):
            chunk_df = pd.read_parquet(chunk_output_path)
            chunks = chunk_df.to_dict(orient='records')
            print("File [chunks.parquet] đã tồn tại. Không ghi đè.")
        else:
            corpus_df = pd.read_parquet(corpus_path)
            corpus_ds = corpus_df.to_dict(orient='records')

            print(f"RAM before: {psutil.virtual_memory().used / 1e9:.2f} GB")
            print(f"VRAM before: {torch.cuda.memory_allocated() / 1e9:.2f} GB")

            chunks = []
            chunk_start_idx = 0
            for c in corpus_ds:
                sample_text = c["context"]
                sample_chunks = self.data_processor.process_text_to_chunks(sample_text, chunk_size, chunk_start_idx)
                chunk_start_idx += len(sample_chunks)
                print(chunk_start_idx)
                chunks.extend(sample_chunks)

            print(f"RAM after: {psutil.virtual_memory().used / 1e9:.2f} GB")
            print(f"VRAM after: {torch.cuda.memory_allocated() / 1e9:.2f} GB")
            print(f"Số chunks: {len(chunks)}")

            chunk_table = pa.Table.from_pylist(chunks)
            pq.write_table(chunk_table, chunk_output_path)

            print("Ghi kết quả ra file [chunks.parquet] thành công.")
            print("Chunking văn bản đầu vào thành công!")


    def er_extract(self, chunk_input_path, er_extracted_chunk_output_path, batch_size=16):
        self._require_llm("er_extract")
        if os.path.exists(er_extracted_chunk_output_path):
            print("File [er_extracted_chunks.parquet] đã tồn tại. Không ghi đè.")
        else:
            rows = []
            if not os.path.exists(chunk_input_path):
                print("File [chunks.parquet] không tồn tại. Hãy thử lại")
                return
            chunk_df = pd.read_parquet(chunk_input_path)
            chunks = chunk_df.to_dict(orient='records')

            for i in tqdm(range(0, len(chunks), batch_size), desc="ER Extraction"):
                batch_chunks = chunks[i:i+batch_size]
                chunk_texts = [c["text"] for c in batch_chunks]
                chunk_indices = [c["chunk_index"] for c in batch_chunks]

                raw_responses = rag.extractor.extract_batch(chunk_texts, chunk_indices)
                results = rag.extractor.parse_batch(raw_responses, chunk_indices)

                for j, res in enumerate(results):
                    row = batch_chunks[j].copy()
                    row["entities"] = res["entities"]
                    row["relations"] = res["relations"]
                    row["content_keywords"] = res["content_keywords"]

                    rows.append(row)

                if (i // batch_size) % 100 == 0:
                    import gc; gc.collect()

            er_df = pd.DataFrame(rows)

            for col in ["entities", "relations", "content_keywords"]:
                for i, val in enumerate(er_df[col]):
                    chunks[i][col] = val
            # Có trường bị thiếu do lỗi khi sinh của LLM
            # Giải pháp nhanh: xóa các quan hệ này đi
            new_chunks = [
                {
                    **chunk,
                    'relations': [
                        rel for rel in chunk['relations']
                        if all(v not in (None, "", []) for v in rel.values())
                    ]
                }
                for chunk in chunks
            ]
            chunk_table = pa.Table.from_pylist(new_chunks)
            pq.write_table(chunk_table, er_extracted_chunk_output_path)

            print("Ghi kết quả ra file [er_extracted_chunks.parquet] thành công.")
            print("Trích xuất thực thể - quan hệ thành công!")

    def er_validate(self, er_extracted_chunk_input_path, validated_er_extracted_chunk_output_path):
        if os.path.exists(validated_er_extracted_chunk_output_path):
            print("File [validated_er_extracted_chunks.parquet] đã tồn tại. Không ghi đè.")
        else:
            if not os.path.exists(er_extracted_chunk_input_path):
                print("File [er_extracted_chunks.parquet] không tồn tại. Hãy thử lại")
                return
            chunk_df = pd.read_parquet(er_extracted_chunk_input_path)
            chunks = chunk_df.to_dict(orient="records")

            for chunk in chunks:
                valid_entity_ids = []
                new_entities = []
                new_relations = []
                for ent in chunk["entities"]:
                    if self.validator.is_valid_entity(ent):
                        valid_entity_ids.append(ent["id"])
                        new_entities.append(ent)
                for rel in chunk["relations"]:
                    if self.validator.is_valid_relation(rel, valid_entity_ids):
                        new_relations.append(rel)
                chunk["entities"] = new_entities
                chunk["relations"] = new_relations
            new_table = pa.Table.from_pylist(chunks)
            pq.write_table(new_table, validated_er_extracted_chunk_output_path)

            print("Ghi kết quả ra file [validated_er_extracted_chunks.parquet] thành công.")
            print("Xác nhận thực thể - quan hệ thành công!")


    def er_deduplicate(self, validated_er_extracted_chunk_input_path, canonical_entity_output_path, final_relation_output_path):
        if not os.path.exists(validated_er_extracted_chunk_input_path):
            print("File [validated_er_extracted_chunks.parquet] không tồn tại. Hãy thử lại")
            return
        if os.path.exists(canonical_entity_output_path):
            print("File [canonical_entities.parquet] đã tồn tại. Không ghi đè.")
            return
        if os.path.exists(final_relation_output_path):
            print("File [final_relations.parquet] đã tồn tại. Không ghi đè.")
            return
        df = pd.read_parquet(validated_er_extracted_chunk_input_path)
        chunks = df.to_dict(orient="records")
        self.deduplicator.group_by_type(chunks)
        self.deduplicator.build_embeddings_pairs()
        self.deduplicator.deduplicate_entity()
        self.deduplicator.build_canonical_lookup()
        self.deduplicator.deduplicate_relation(chunks)

        canonical_entity_table = pa.Table.from_pylist(self.deduplicator.canonical_entities)
        pq.write_table(canonical_entity_table, canonical_entity_output_path)

        final_relation_table = pa.Table.from_pylist(self.deduplicator.final_relations)
        pq.write_table(final_relation_table, final_relation_output_path)

        print("Ghi kết quả ra file [canonical_entities.parquet] thành công.")
        print("Ghi kết quả ra file [final_relations.parquet] thành công.")
        print("Loại bỏ trùng lặp thực thể thành công!")



    def build_kg(self, canonical_entity_input_path, final_relation_input_path):
        if not os.path.exists(canonical_entity_input_path):
            print("File [canonical_entities.parquet] không tồn tại. Hãy thử lại")
            return
        if not os.path.exists(final_relation_input_path):
            print("File [final_relations.parquet] không tồn tại. Hãy thử lại")
            return

        entity_df = pd.read_parquet(canonical_entity_input_path)
        canonical_entities = entity_df.to_dict(orient="records")

        relation_df = pd.read_parquet(final_relation_input_path)
        final_relations = relation_df.to_dict(orient="records")

        self.graph_store.add_nodes(canonical_entities)
        self.graph_store.add_edges(final_relations)

        print("Tạo đồ thị tri thức thành công!")

    def build_vector_store(self, canonical_entity_input_path, final_relation_input_path):
        if not os.path.exists(canonical_entity_input_path):
            print("File [canonical_entities.parquet] không tồn tại. Hãy thử lại")
            return
        if not os.path.exists(final_relation_input_path):
            print("File [final_relations.parquet] không tồn tại. Hãy thử lại")
            return

        entity_df = pd.read_parquet(canonical_entity_input_path)
        canonical_entities = entity_df.to_dict(orient="records")

        relation_df = pd.read_parquet(final_relation_input_path)
        final_relations = relation_df.to_dict(orient="records")

        self.vector_store.build_entity_store(canonical_entities)
        self.vector_store.build_relation_store(final_relations)

        print("Tạo vector store thành công!")


    def extract_keywords_from_query(self, question_path, keywords_extracted_question_output_path, batch_size=16):
        self._require_llm("extract_keywords_from_query")
        if os.path.exists(keywords_extracted_question_output_path):
            print("File [kw_extracted_questions.parquet] đã tồn tại. Không ghi đè.")
            return
        if not os.path.exists(question_path):
            print("File [(name)_questions.parquet] không tồn tại. Hãy thử lại")
            return
        question_df = pd.read_parquet(question_path)
        question_ds = question_df.to_dict(orient='records')
        querys = [question['question'] for question in question_ds]

        raw_responses = []
        for i in tqdm(range(0, len(querys), batch_size), desc="Query Seed Entity Extraction"):
            batch_querys = querys[i:i+batch_size]
            batch_raw_responses = self.seed_entity_extractor.extract_batch(batch_querys)
            raw_responses.extend(batch_raw_responses)
            if (i // batch_size) % 100 == 0:
                gc.collect()
        extracted_entities = self.seed_entity_extractor.parse_batch(raw_responses)

        question_infos = []

        """
        Chuẩn bị dữ liệu để đánh giá retrieval
        
        Đầu vào để đánh giá cần có:
        
        retrieval_eval_data = {
            "id": ...,
            "source": ...,
            "question": ...,
            "question_type": ...,
            "evidence": ...,
            "context": ...,
        }
        
        """
        for q, entities in zip(question_ds, extracted_entities):
            question_infos.append({'id': q['id'], 
                                   'source': q['source'], 
                                   'question': q['question'], 
                                   'seed_entities': entities['seed_entities'], 
                                   'question_type': q['question_type'], 
                                   'evidence': q['evidence'].split(';'),

            })

        kw_extracted_question_table = pa.Table.from_pylist(question_infos)
        pq.write_table(kw_extracted_question_table, keywords_extracted_question_output_path)

        print("Ghi kết quả ra file [kw_extracted_questions.parquet] thành công.")
        print("Trích xuất keyword của các câu truy vấn thành công!")


    def retrieve_context_for_answering(
        self,
        seed_entities_question_input_path: str,
        context_retrieved_question_output_path: str,
        validated_er_extracted_chunk_input_path: str,
        top_k,
        config: Dict = None,
           
    ):
        """
        Retrieve context bằng Graph PPR:
          seed_entities text
          -> vector entity search
          -> weighted graph seed nodes
          -> connected components
          -> nếu seed node isolated:
               lấy chunk chứa isolated node
               lấy thêm chunk liền trước/liền sau của chunk đó
          -> weighted PPR nếu có connected graph
          -> rerank
          -> final top_k chunks
        
        Không dùng E1/E2/E3.
        Không dùng dense chunk retrieval.
        Không dùng BM25 fallback.
        Fallback chỉ dùng local neighbor chunks quanh isolated seed node.
        """
            
        if not os.path.exists(seed_entities_question_input_path):
            print("File [kw_extracted_questions.parquet] không tồn tại.")
            return
    
        if not os.path.exists(validated_er_extracted_chunk_input_path):
            print("File [validated_er_extracted_chunks.parquet] không tồn tại.")
            return
    
        if os.path.exists(context_retrieved_question_output_path):
            print(f"File [{context_retrieved_question_output_path}] đã tồn tại. Không ghi đè.")
            return
    
        question_df = pd.read_parquet(seed_entities_question_input_path)
        question_ds = question_df.to_dict(orient="records")
        

        graph_config = {
            "top_k": top_k,
        
            # Seed resolution: giữ tương đối chặt, không hạ threshold quá thấp
            "seed_resolver_top_k": 5,
            "seed_resolver_threshold": 0.65,
            "seed_resolver_fallback_top_k": 1,
            "use_vector_seed_resolution": True,
            "use_vector_scores_as_seed_weights": True,
        
            # PPR
            "top_ppr_nodes": 30,
            "alpha": 0.6,
            "max_iter": 50,
            "convergence_eps": 1e-6,
        
            # Component/subgraph
            "resource_threshold": 1e-4,
            "max_component_size": 800,
            "fallback_hop_radius": 1,
        
            # Isolated node direct score
            "direct_isolated_score": 1.0,
        
            # Isolated node neighbor fallback
            # Với mỗi isolated node:
            #   chunk_index = anchor
            #   lấy anchor-2, anchor-1, anchor, anchor+1, anchor+2
            "use_isolated_neighbor_fallback": True,
            "isolated_neighbor_before": 2,
            "isolated_neighbor_after": 2,
            "isolated_neighbor_decay": 0.85,
            "isolated_neighbor_score_ratio": 0.8,
        
            # Path: tắt trước để giảm nhiễu
            "use_paths": False,
            "path_cutoff": 5,
            "top_paths": 20,
            "max_path_per_pair": 5,
            "path_search_timeout": 10.0,
        
            # Graph score
            "weight_ppr_node": 1.0,
            "weight_path": 0.0,
        
        
            # Rerank
            "use_reranker": True,
            "candidate_multiplier": 4,
            "min_candidate_count": 30,
            "rerank_weight": 0.85,
            "graph_weight_after_rerank": 0.15,
        }
    
        if config:
            graph_config.update(config)
    
        self.retriever = Retriever(
            graph_store=self.graph_store,
            vector_store=self.vector_store,
            chunks=chunks,
            reranker=self.reranker,
            config=graph_config,
        )
    
        contexts = []
    
        for i, question_record in enumerate(
            tqdm(question_ds, desc="Graph PPR Retrieval")
        ):
            question = question_record.get("question", "")
            seed_entities = question_record.get("seed_entities", [])
            q_id = question_record.get("id", "")
            source = question_record.get("source", "")
            question_type = question_record.get("question_type", "")
            evidence = question_record.get("evidence", [])
            
    
            if isinstance(seed_entities, np.ndarray):
                seed_entities = seed_entities.tolist()
    
            if not seed_entities:
                print(f"[WARNING] Question {i}: No seed entities extracted")
                contexts.append({
                    "id": q_id,
                    "source": source,
                    "question": question,
                    "question_type": question_type,
                    "evidence": evidence,
                    "context": [],
                    "retrieved_chunks": [],
                    "retrieval_details": {
                        "mode": "graph_ppr",
                        "seed_entities": [],
                        "warning": "No seed entities extracted",
                    },
                })
                continue
    
            try:
                result = self.retriever.retrieve(
                    seed_entities=seed_entities,
                    question=question,
                    top_k=top_k,
                )
    
                final_chunks = result.get("final_chunks", [])
    
                chunk_context = [
                    c.get("text")
                    for c in final_chunks
                    if c.get("text", "") and len(c.get("text", "")) != 0
                ]
    
                contexts.append({
                    "id": q_id,
                    "source": source,
                    "question": question,
                    "question_type": question_type,
                    "evidence": evidence,
                    "context": chunk_context,
                    "retrieved_chunks": [
                        {
                            "chunk_index": c.get("chunk_index"),
                            "final_score": c.get("final_score"),
                            "rerank_score": c.get("rerank_score"),
                            "rerank_score_norm": c.get("rerank_score_norm"),
                            "graph_score": c.get("graph_score"),
                            "graph_score_norm": c.get("graph_score_norm"),
                        }
                        for c in final_chunks
                    ],
                    "retrieval_details": {
                        "mode": "graph_ppr_vector_seed_resolution",
                        "seed_entities": seed_entities,
                        "graph_scores": [
                            {"chunk_id": int(chunk_id), "score": float(score)}
                            for chunk_id, score in result.get("graph_scores", {}).items()
                        ],
                        **result.get("retrieval_details", {}),
                    },
                })

    
            except Exception as e:
                print(f"[ERROR] Question {i}: {e}")
                traceback.print_exc()
                contexts.append({
                    "id": q_id,
                    "source": source,
                    "question": question,
                    "question_type": question_type,
                    "evidence": evidence,
                    "context": [],
                    "retrieved_chunks": [],
                    "retrieval_details": {
                        "mode": "graph_ppr_vector_seed_resolution",
                        "seed_entities": seed_entities,
                        "error": str(e),
                    },
                })
    
            if (i + 1) % 100 == 0:
                gc.collect()
    
        pq.write_table(
            pa.Table.from_pylist(contexts),
            context_retrieved_question_output_path,
        )
    
        print(f"Ghi kết quả ra file [{context_retrieved_question_output_path}] thành công.")
        print("Trích xuất context bằng Graph PPR thành công!")

    def question_answering(self, context_retrieved_question_input_path, question_path, completed_question_answer_output_path, max_context_tokens=9000, batch_size=16):
        self._require_llm("question_answering")
        if not os.path.exists(context_retrieved_question_input_path):
            print("File [context_retrieved_questions.parquet] không tồn tại. Hãy thử lại")
            return
        if not os.path.exists(question_path):
            print("File [(name)_questions.parquet] không tồn tại. Hãy thử lại")
            return
        if os.path.exists(completed_question_answer_output_path):
            print("File [answered_questions.parquet] đã tồn tại. Không ghi đè.")
            return

        if max_context_tokens:
            self.answer_generator.max_context_tokens = max_context_tokens

        context_df = pd.read_parquet(context_retrieved_question_input_path)
        context_ds = context_df.to_dict(orient='records')

        raw_answers = []
        effective_contexts_all = []
        
        for i in tqdm(range(0, len(context_ds), batch_size), desc="Question Answering"):
            batch_contexts = context_ds[i:i+batch_size]
        
            responses, effective_contexts = self.answer_generator.answer_batch(batch_contexts)
        
            answers = self.answer_generator.parse_batch(responses)
        
            raw_answers.extend(answers)
            effective_contexts_all.extend(effective_contexts)
        
            if (i // batch_size) % 100 == 0:
                gc.collect()

        question_df = pd.read_parquet(question_path)
        question_ds = Dataset.from_pandas(question_df)
        question_ds = question_ds.add_column("rag_answer", raw_answers)
        
        question_ds = question_ds.add_column("effective_context", effective_contexts_all)

        question_ds.to_parquet(completed_question_answer_output_path)

        print("Ghi kết quả ra file [answered_questions.parquet] thành công.")
        print("Đã trả lời xong các câu hỏi!")

    question_types = {"Fact Retrieval" : 0, "Complex Reasoning" : 1, "Contextual Summarize": 2, "Creative Generation": 3}
````

````python
print(f"Total canonical_entities: {len(canonical_entities)}")
print(f"Total final_relations: {len(final_relations)}")

# Lấy tất cả entity_id
entity_ids = set(e['self']['id'] for e in canonical_entities)

# Lấy tất cả id xuất hiện trong relations
relation_source_ids = set(r['source_id'] for r in final_relations)
relation_target_ids = set(r['target_id'] for r in final_relations)
relation_ids = relation_source_ids | relation_target_ids

# Entities không xuất hiện trong bất kỳ relation nào
never_in_relation = entity_ids - relation_ids

# Relation trỏ đến entity không tồn tại
ghost_ids = relation_ids - entity_ids

print(f"Entities không có relation nào: {len(never_in_relation)}")
print(f"Relation trỏ đến entity không tồn tại (ghost): {len(ghost_ids)}")

# Xem thử vài isolated entity trông như thế nào
for eid in list(never_in_relation)[:5]:
    ent = next(e for e in canonical_entities if e['self']['id'] == eid)
    print(f"  → {ent}")
````

````python
degrees = defaultdict(int)

for rel in final_relations:
    degrees[rel["source_id"]] += 1
    degrees[rel["target_id"]] += 1

isolated = 0

for ent in canonical_entities:
    if degrees[ent["self"]["id"]] == 0:
        isolated += 1

print(isolated)
````

````python
import numpy as np

deg_values = list(degrees.values())

print("mean =", np.mean(deg_values))
print("median =", np.median(deg_values))
print("max =", np.max(deg_values))
````

````python
import pandas as pd
import json
import numpy as np
import datetime

def json_default(o):
    # numpy array
    if isinstance(o, np.ndarray):
        return o.tolist()

    # numpy scalar
    if isinstance(o, np.integer):
        return int(o)

    if isinstance(o, np.floating):
        if np.isnan(o):
            return None
        return float(o)

    if isinstance(o, np.bool_):
        return bool(o)

    # pandas missing values
    if pd.isna(o):
        return None

    # datetime
    if isinstance(o, (pd.Timestamp, datetime.datetime, datetime.date)):
        return o.isoformat()

    # fallback
    return str(o)

df = pd.read_parquet(context_retrieved_question_input_path)
df_records = df[['id', 'source', 'question', 'question_type', 'evidence', 'context']].to_dict(orient="records")

context_json_path = "/kaggle/working/retrieval_eval_data.json"

with open(context_json_path, "w", encoding="utf-8") as f:
    json.dump(
        df_records,
        f,
        ensure_ascii=False,
        indent=2,
        default=json_default  
    )

print("Done")
````

````python
"""
Chuẩn bị dữ liệu để sinh câu trả lời 

Đầu vào để đánh giá cần có:

generation_input_data = {
    "id": ...,
    "source": ...,
    "question": ...,
    "question_type": ...,
    "answer": ...,
    "evidence": ...,
    "context": ...,
}
Khi sử dụng cái này đế làm đầu vào sinh câu trả lời rồi thi đầu ra của phần này sẽ tự khớp với đầu vào của pha đánh giá hiệu năng sinh câu trả lời
"""
````

````python
import pandas as pd
import json
import numpy as np
import datetime

def json_default(o):
    # numpy array
    if isinstance(o, np.ndarray):
        return o.tolist()

    # numpy scalar
    if isinstance(o, np.integer):
        return int(o)

    if isinstance(o, np.floating):
        if np.isnan(o):
            return None
        return float(o)

    if isinstance(o, np.bool_):
        return bool(o)

    # pandas missing values
    if pd.isna(o):
        return None

    # datetime
    if isinstance(o, (pd.Timestamp, datetime.datetime, datetime.date)):
        return o.isoformat()

    # fallback
    return str(o)

context_df = pd.read_parquet(context_retrieved_question_input_path)
question_df = pd.read_parquet(question_path)
generation_input_df = context_df.copy()
generation_input_df['answer'] = question_df['answer']
generation_input_records = generation_input_df[['id', 'source', 'question', 'question_type', 'answer', 'evidence', 'context']].to_dict(orient="records")

generation_json_path = "/kaggle/working/generation_input_data.json"

with open(generation_json_path, "w", encoding="utf-8") as f:
    json.dump(
        generation_input_records,
        f,
        ensure_ascii=False,
        indent=2,
        default=json_default   # Quan trọng nhất là dòng này
    )

print("Done")
````
